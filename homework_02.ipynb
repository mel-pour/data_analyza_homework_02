```python
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
btc_df = yf.download("BTC-USD", start="2018-01-01", end="2026-11-23", multi_level_index=False)
btc_df.head()
```

```python
btc_df["Benefit"] = btc_df["Close"] - btc_df["Open"]
btc_df.head()
```

```python
gold_df = yf.download("GC=F", start="2018-01-01", end="2026-11-23", multi_level_index=False)
gold_df.head()
```

```python
oil_df = yf.download("CL=F", start="2018-01-01", end="2026-11-23", multi_level_index=False)
oil_df.head()
```

```python
usd_df = yf.download("EURUSD=x", start="2018-01-01", end="2026-11-23", multi_level_index=False)
usd_df.head()
```

```python
plt.figure(figsize=(16,6))
plt.plot(btc_df["Close"],label='Bitcoin')
plt.plot(gold_df["Close"],label='GOLD')
plt.plot(oil_df["Close"],label='OIL')
plt.plot(usd_df["Close"],label='USD')
plt.legend()

plt.legend()
plt.show()
```

#all four without scaling

```python
plt.figure(figsize=(16,6))

plt.subplot(1,2,1)
plt.title("BTC")
plt.plot(btc_df["Close"])

plt.subplot(1,2,2)
plt.title("USD")
plt.plot(usd_df["Close"])
```

```python
plt.figure(figsize=(16,6))

plt.subplot(1,2,1)
plt.title("GOLD")
plt.plot(gold_df["Close"])

plt.subplot(1,2,2)
plt.title("OIL")
plt.plot(oil_df["Close"])
```

```python
btc_close_scaled = (btc_df["Close"] - btc_df["Close"].mean()) / btc_df["Close"].std()
# scaler = StandardScaler()
# btc_close_scaled = scaler.fit_transform(btc_df["Close"])
btc_close_scaled.mean(),btc_close_scaled.std()
```

```python
gold_close_scaled = (gold_df["Close"] - gold_df["Close"].mean()) / gold_df["Close"].std()
gold_close_scaled.mean(),gold_close_scaled.std()
```

```python
oil_close_scaled = (oil_df["Close"] - oil_df["Close"].mean()) / oil_df["Close"].std()
oil_close_scaled.mean(),oil_close_scaled.std()
```

```python
usd_close_scaled = (usd_df["Close"] - usd_df["Close"].mean()) / usd_df["Close"].std()
# eth_close_scaled = scaler.fit_transform(eth_df["Close"])
usd_close_scaled.mean(), usd_close_scaled.std()
```

```python
plt.figure(figsize=(16,6))

plt.plot(btc_close_scaled, label='Bitcoin')
plt.plot(gold_close_scaled, label='Gold')
plt.plot(oil_close_scaled, label='Oil')
plt.plot(usd_close_scaled, label='USD')
plt.legend()

plt.legend()
plt.show()
```
